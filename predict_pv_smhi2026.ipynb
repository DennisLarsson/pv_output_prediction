```python
import pandas as pd
import numpy as np

from Solar.regression.process_smhi import process_smhi

smhi_filename = "../data/smhi_Gi_Jan01-2026-May31-2026.txt"

X_smhi_2026, raw_smhi_2026 = process_smhi(smhi_filename)
X_smhi_2026
```

```python
raw_smhi_2026
```

```python
import joblib

fitted_gb_reg = joblib.load('../models/fitted_gb_reg.joblib')
```

```python
smhi_2026_pred = fitted_gb_reg.predict(X_smhi_2026)

for line in smhi_2026_pred:
    print(line)
```

```python
smhi_2026_pred_clipped = np.where(smhi_2026_pred < 0.1, 0, smhi_2026_pred)
for line in smhi_2026_pred_clipped:
    print(line)
```

```python
smhi_2026_time = raw_smhi_2026['datetime'].copy()
type(smhi_2026_time)
```

```python
result = pd.DataFrame({
    'time': smhi_2026_time,
    'predicted_pv_output': smhi_2026_pred_clipped
})
print(result)
```

```python
result.to_csv("smhi_2026_predicted_pv_output_with_time.csv", index=False)
```
